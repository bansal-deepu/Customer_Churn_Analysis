# Feature engineering

## Customer_Value

```python
df['Customer_Value'] = df.Tenure_Months * df.Monthly_Charges
```

```python
df.head(2)
```

## Tenure_Groups

```python
bins = [0, 12, 24, 48, 72]
labels = ["0-12 month", "13-24 month", "25-48 month", "49-72 month"]

df['Tenure_Group'] = pd.cut(
    df['Tenure_Months'], 
    bins=bins, 
    labels=labels, 
    include_lowest=True
)
```

```python
df['Tenure_Group'].value_counts(dropna=False)
```

## Age_Group

```python
df['Age_Group'] = np.where(df['Age'] >= 60, 'Senior', np.where(df['Age'] >= 30, 'Adult', 'Young'))
```

```python
df['Age_Group'].value_counts(dropna=False)
```

## Churn_Flag

```python
df['Churn'].value_counts(dropna=False)
```

```python
df['Churn_Flag'] = df['Churn'].map({'No':0,'Yes':1})
```

```python
df['Churn_Flag'].value_counts(dropna=False)
```
